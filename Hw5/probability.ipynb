```python
from probability import *
```

```python
T, F = True, False

cancer = BayesNet([
    ('Smoker', '', 0.30),                                  
    ('Cancer', 'Smoker', {T: 0.05, F: 0.01}),               
    ('XRay', 'Cancer', {T: 0.90, F: 0.20})                  
])

cancer
```

ข้อ 41

```python
p_s = cancer.variable_node('Smoker').p(T, {})
p_c_given_s = cancer.variable_node('Cancer').p(T, {'Smoker': T})
p_t_given_c = cancer.variable_node('XRay').p(T, {'Cancer': T})

ans1 = p_s * p_c_given_s * p_t_given_c
print("ข้อ 1: P(S=true, C=true, T=บวก) =", round(ans1, 4))
```

```python
full_joint = JointProbDist(['Smoker', 'Cancer', 'XRay'])

for s in [T, F]:
    ps = cancer.variable_node('Smoker').p(s, {})
    for c in [T, F]:
        pc = cancer.variable_node('Cancer').p(c, {'Smoker': s})
        for x in [T, F]:
            px = cancer.variable_node('XRay').p(x, {'Cancer': c})
            full_joint[dict(Smoker=s, Cancer=c, XRay=x)] = ps * pc * px

full_joint
```

ข้อ 42

```python
evidence = dict(XRay=T)
variables = ['Smoker', 'Cancer']  
ans42 = enumerate_joint(variables, evidence, full_joint)
print("ข้อ 42: P(T=บวก) =", round(ans42, 4))
```

ข้อ 43

```python
ans43_dist = enumerate_joint_ask('Cancer', dict(XRay=T), full_joint)
print("ข้อ 43: P(C=true | T=บวก) =", round(ans43_dist[T], 4))
```

ข้อ 44

```python
ans44_dist = enumerate_joint_ask('Smoker', dict(XRay=T), full_joint)
print("ข้อ 44: P(S=true | T=บวก) =", round(ans44_dist[T], 4))
```

ข้อ 45

```python
evidence_45 = dict(Cancer=T, XRay=T)
variables_45 = ['Smoker'] 
p_c_and_t = enumerate_joint(variables_45, evidence_45, full_joint)
ans45 = 1000 * p_c_and_t
print("ข้อ 45: จำนวนคนที่เป็นมะเร็งและผลบวก ≈", round(ans45, 1))
```
